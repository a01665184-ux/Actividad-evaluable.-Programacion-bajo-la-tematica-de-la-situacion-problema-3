```python
from IPython.display import HTML
from base64 import b64encode

def play_video(filename):
    # Abre el archivo de video y lo codifica en base64 para que el navegador lo pueda reproducir directamente
    video_data = open(filename, "rb").read()
    encoded = b64encode(video_data).decode('ascii')
    return HTML(f'''<video width="400" controls>
        <source src="data:video/mp4;base64,{encoded}" type="video/mp4">
    </video>''')

# Reproducir el video original
play_video('gloria_bonder.mp4')
```

##Comandos

-f fmt (input/output)
Force input or output file format. The format is normally auto detected for input files and guessed from the file extension for output files, so this option is not needed in most cases.

-i url (input)
input file url

-y (global)
Overwrite output files without asking.

-n (global)
Do not overwrite output files, and exit immediately if a specified output file already exists.

-stream_loop number (input)
Set number of times input stream shall be looped. Loop 0 means no loop, loop -1 means infinite loop.

-recast_media (global)
Allow forcing a decoder of a different media type than the one detected or designated by the demuxer. Useful for decoding media data muxed as data streams.

-c[:stream_specifier] codec (input/output,per-stream)
-codec[:stream_specifier] codec (input/output,per-stream)
Select an encoder (when used before an output file) or a decoder (when used before an input file) for one or more streams. codec is the name of a decoder/encoder or a special value copy (output only) to indicate that the stream is not to be re-encoded.

```python
import cv2

# Correct path to the available video file
videofile = '/content/prueba.mp4'

# Load the uploaded video
cap = cv2.VideoCapture(videofile)

# Get the video parameters
frame_width = int(cap.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(cap.get(cv2.CAP_PROP_FRAME_HEIGHT))
fps = int(cap.get(cv2.CAP_PROP_FPS)) if cap.get(cv2.CAP_PROP_FPS) > 0 else 30

# Use 'mp4v' codec to output mp4 video file
fourcc = cv2.VideoWriter_fourcc(*'mp4v')

# Web players display mp4 files better when written with 3-channels (isColor=True)
output = cv2.VideoWriter('processed_video.mp4', fourcc, fps, (frame_width, frame_height), isColor=True)

while cap.isOpened():
    ret, frame = cap.read()
    if not ret:
        break

    # Apply transformations (Grayscale)
    gray_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)

    # Convert back to BGR 3-channel so the video encoder writes it correctly
    gray_3channel = cv2.cvtColor(gray_frame, cv2.COLOR_GRAY2BGR)

    # Write the processed frame to output
    output.write(gray_3channel)

cap.release()
output.release()
print("Grayscale video processed and saved successfully as 'processed_video_prueba.mp4'.")
```
